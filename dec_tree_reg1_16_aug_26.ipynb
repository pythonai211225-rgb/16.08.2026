```python
# MSE
# Mean Squared Error
# Mean + MSE 
def mse_avg(arr):
    return np.mean((arr - arr.mean())**2)

y = np.array([     10,     12,     25,     24,     40,     42])
X = np.array([ [1, 3], [2, 1], [3, 2], [4, 3], [5, 1], [6, 2]])
print('root')
mse_root = mse_avg(y)
print(' MSE:', mse_root, 'AVG:', y.mean())
print(' x1 <= 3.5')

y_yes = np.array([10, 12, 25])
y_yes_mse = mse_avg(y_yes)
print('  Yes', y_yes, y_yes_mse)

y_no = np.array([24, 40, 42])
y_no_mse = mse_avg(y_no)
print('  No ', y_no,  y_no_mse)      # 60 2   80 4 

weighted_mse = y_yes_mse * 3/6 + y_no_mse * 3/6
print('  weighted mse?', weighted_mse)
print('  mse gain?', mse_root - weighted_mse)

print()
print('  ', y_yes)
print('  YES: x1 <= 2')
y_yes2 = np.array([10, 12])
y_yes_mse2 = mse_avg(y_yes2)
print('  Yes', y_yes2, y_yes_mse2)

y_no3 = np.array([25])
y_no_mse2 = mse_avg(y_no3)
print('  No ', y_no3,  y_no_mse2)      # 60 2   80 4 

weighted_mse2 = y_yes_mse2 * 2/3 + y_no_mse2 * 1/3
print('  weighted mse?', weighted_mse2)
print('  mse gain?', weighted_mse - weighted_mse2)
```

```python
from sklearn.tree import DecisionTreeRegressor, export_text
from sklearn.metrics import r2_score

X = np.array( [1, 2, 3, 4, 5, 6]).reshape(-1, 1)
y = np.array( [1, 2, 3, 8, 9, 10])

tree = DecisionTreeRegressor(criterion='squared_error', max_depth=1, random_state=42)
tree.fit(X, y)

y_pred = tree.predict(X)

print(X.ravel())
print("y true: ", y)
print("y pred: ", y_pred)

tree_rules = export_text(tree, feature_names=["x"])
print(tree_rules)

print('Performance R²', r2_score(y, y_pred), 'r2_score')
```
